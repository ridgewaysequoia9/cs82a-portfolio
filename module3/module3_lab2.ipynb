```python
import pandas as pd

# Started by loading the CSV file into a table called 'df'
df = pd.read_csv('usarrests.csv')

# Going to give the first columna clear name instead of 'Unnamed: 0'
df = df.rename(columns={'Unnamed: 0': 'State'})

# Show the first five rows so we know it loaded properly 
df.head()
```

```python
# Getting the summary statistics (count, mean, std, min, quartiles, max)
df.describe()
```

```python
# Showing the row that has the missing value (NaN)
print("Missing Value:")
display(df[df.isnull().any(axis=1)])

# Show rows were UrbanPop is over 100% or unusually low (under 20%)
print("Suspicious UrbanPop:")
display(df[(df['UrbanPop'] > 100) | (df['UrbanPop'] < 20)])

# Rows were assault is unusually high (over 400)
print("Suspicious Assault:")
display(df[(df['Assault'] > 400)])

```

### Flagged Values: Verdicts and Cleaning Log
- Iowa(UrbanPop = 570): Iowa population was listed as 570, which I knew was incorrect because a percentage can only go up to 100%. So since that number was clearly a mistake, I replaced 570 with NaN so I could keep Iowa's valid murder and assault numbers without deleting the whole state.
- New York(UrbanPop = 6): New York showed an urban population of 6%, which was probably a typo, which could be a mathematically  impossible since 6 is between 0 and 100. But even though 6% seem way to low for New York, i just left it at 6 and flagged it since I dont have the original source to confirm the real number.
- South Carolina(Assault = 879.0): South Carolina has an assault rate of 879.0 per 100,000 people, which seemed extreme compared to every other state. Crime rates aren't capped at 100 like a percentage, it's not impossible but just suspiciously high, so I just kept it in the dataset and noted as a major outlier.
- Georgia(Assault = NaN): Georgia was missing the assualt number, which is the reason why the count in describe() was 49 instead of 50. It was missing value rather then just an impossible or extreme number, so I decided to leave it as NaN instead of deleting Georgia's row so we don't lose the valid murder 17.4 and urban population 60 data.  

```python
import numpy as np

# Replacing Iowa's impossible UrbanPop value (570) with NaN
df.loc[df['State'] == 'Iowa', 'UrbanPop'] = np.nan

# Re-run describe() to show the cleaned summary statistics
df.describe()
```
