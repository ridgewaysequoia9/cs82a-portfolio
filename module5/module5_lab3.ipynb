```python
import pandas as pd

# Loading the tab-seperated file
df = pd.read_csv('custdata.tsv', sep='\t')

# Select the three columns for Problem 5.5
cols = ['age', 'income', 'num.vehicles']

# Quick check of summary statistics to spot impossible values
print("Total rows:", len(df))
display(df[cols].describe())
print("\nMissing values per column:")
print(df[cols].isna().sum())
```

```python
corr_before = df[cols].corr()
print("Correlation Matrix BEFORE Cleaning:")
display(corr_before.round(4))
```

```python
print("Ages <= 0:",  (df['age'] <= 0).sum())
print("Ages > 100:", (df['age'] > 100).sum())
print("Negative income:", (df['income'] < 0).sum())
print("Missing values:", df[cols].isna().any(axis=1).sum())
```

```python
df_clean = df[
    (df['age'] > 0) &
    (df['age'] <= 100) &
    (df['income'] >= 0) &
    (df['num.vehicles'].notna())
    ].copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Total removed:", len(df)  - len(df_clean))

corr_after = df_clean[cols].corr()
print("\nCorrelation Matrix AFTER Cleaning:")
display(corr_after.round(4))
```

So first, to avoid invalid records from distorting the correlation results, I removed **68** rows from the original **1,000** records, leaving **932** valid rows: 
* I removed **3** rows with `age <= 0` and **8** rows with `age > 100`, as these are unrealistic ages that skew the the data. 
* Then one row was removed with `income < 0`, since negative income is an invalid entry for this dataset.
* Finally, removed **56** rows containing missing values in `num_vehicles` so all three correlations were calculated on the same set of completed customers.

* Then after cleaning, **income and number of vehicles** has the strongest relationship at **$r = 0.1379$**, which is still a very weak positive correlation higher income, customers tend to own only a slight more vehicles. **Age and number of vehicles** ($r = -0.0663$) and ($r = -0.0026$) show virtually no linear relationship. But overall, none of the three variables predict one another in thge cleaned dataset.
