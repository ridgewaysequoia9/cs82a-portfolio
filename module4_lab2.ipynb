```python
import pandas as pd

books = pd.read_csv("allbacks.csv")
books
```

```python
sorted_covers = sorted(books["cover"])
median_cover = sorted_covers[len(sorted_covers) // 2]

print("Sorted covers:", sorted_covers)
print("Median covers:", median_cover)
```

```python
mean_weight = books["weight"].mean()
print(f"Overall mean weight: {mean_weight:.2f} g")
```

```python
group_means = books.groupby('cover')['weight'].mean()
print(group_means)
```
