```python
import pandas as pd

bridges = pd.read_csv("bridges.csv")
print(bridges.info())
bridges["length"].describe()
```

### Real-World CutOff Rule (500ft - 3,500ft)

# Plausible Outcome
Before I checked the individual rows, I set a plausible range of 500 feet to 3,500 feet for the length column. These bridges cross major rivers in Pittsburg, a bridge shorter than 500 feet would most likely be to short to span the river channels, while a bridge that is loger than 3,500 feet would be unusually long even if you include approach ramps. Any bridge outside of the range 500 feet to 3,500 feet should be flagged for closer inspection to determine whether it is a data entry error or a valid large scale structure.

```python
min_length = 500
max_length = 3500

flagged_bridges = bridges[(bridges["length"] < min_length) | (bridges["length"] > max_length)]
print(f"Flagged bridges count: {len(flagged_bridges)}")
flagged_bridges
```

```python
# Removing the 3 flagged rows outside the 500-3,500ft range
bridges_clean = bridges[~bridges.index.isin(flagged_bridges.index)].copy()
print("Remaining dataset row count:", len(bridges_clean))
bridges_clean["length"].describe()
```

So due to my 500-3,500ft plausibility rule, three bridges were flagged for exceeding the upper limit. I decided to remove these three rows from the dataset, bringing the total rows from 108 down to 105 now. But even though these are real steel bridges with long spans, their unusual long measurements pulls the average bridge length up from about 1,470 feet to 1,567 feet, so filtering them out prevents a few long span structures from skewing the summary statistic's and keeps the dataset focused on standard-length Pittsburgh bridges.
