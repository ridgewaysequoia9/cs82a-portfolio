```python
import os
import pandas as pd
import matplotlib.pyplot as plt

path = os.path.expanduser('~/Downloads/truck.csv')
df = pd.read_csv(path)

print(df.head())
print(df.describe())
```

```python
r = df['X'].corr(df['Y'])
print("Pearson's r:", round(r, 4))
```

```python
plt.scatter(df['X'], df['Y'])
plt.title('Truck Sale Prices Increase Linearly With List Prices (r = 0.999)')
plt.xlabel('List Price ($ in thousands)')
plt.ylabel('Best Sale Price ($ in thousands)')
plt.show()
```

So for a truck with a list price of 18,000 x = 18.0, I predicted the best sale price of around 15,750 y = 15.75. In the dataset, a list price of 17,900 sold for 15,600 and a list price of 18,400 sold for 16,100, so then 18,000 falls right in between those two points along the same straight trend.
I feel confident in my prediction because Pearson's r is 0.9985, which would be almost 1, and the scatter plot shows all 23 points lined up closely in a straight line. But one limitation is that this dataset only covers trucks priced between 12,400 and 22,400, so the prediction wouldnt be reliable for trucks priced way outside that range.
